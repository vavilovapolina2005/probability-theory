# Задача 2.9

> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене:
> 1) не будут проданы 5 пакетов;
> 2) будет продано:
> а) менее 2 пакетов;
>б) не более 2 пакетов;
>в) хотя бы 2 пакета;
>г) наивероятнейшее число пакетов.

1. не будут проданы 5 пакетов;

```python
import scipy.stats as stats
import math
round(stats.binom.pmf(4, 9, 0.2), 4)
```

2. будет продано: а) менее 2 пакетов;

```python
round(stats.binom.cdf(1, 9, 0.2), 4)
```

2. будет продано: б) не более 2 пакетов;

```python
round(stats.binom.cdf(2, 9, 0.2), 4)
```

2. будет продано: в) хотя бы 2 пакета;

```python
round(1 - stats.binom.cdf(1, 9, 0.2), 4)
```

2. будет продано: г) наивероятнейшее число пакетов.

```python
round(stats.binom.pmf(1, 9, 0.2) + stats.binom.pmf(2, 9, 0.2), 4)
```

# Задача 2.11

> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины:

> а) 480 предприятий;

> б) наивероятнейшее число предприятий;

> в) не менее 480;

> г) от 480 до 520.

а) 480 предприятий;

```python
n = 1000
p = 0.5
s = math.sqrt(n * p * (1 - p))
x = (480 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

б) наивероятнейшее число предприятий;

```python
x = (500 - n * p) / s
round(stats.norm.pdf(x) / s, 4)
```

в) не менее 480;

```python
x1 = (480 - n * p) / s
x2 = (1000 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```

г) от 480 до 520.

```python
x1 = (480 - n * p) / s
x2 = (520 - n * p) / s
round(stats.norm.cdf(x2) - stats.norm.cdf(x1), 4)
```
