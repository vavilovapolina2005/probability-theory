# Задача 1.9

> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, 
> в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

### Решение

* Рассмотрим событие $A$ - составление семизначного числа из цифр с повторениями.

Т. к. цифры повторяются, используем перестановки с повторениями:

$P_n(n_1,n_2,...,n_k)=\frac{n!}{n_1! \cdot n_2! \cdot ... \cdot n_k!}$

```python
import scipy.special as sc
# количество цифр в числе
n = 7
# количество повторений
n1 = 3
n2 = 2
n3 = 2
result = sc.factorial(n) / (
    sc.factorial(n1) *
    sc.factorial(n2) *
    sc.factorial(n3)
)
print("Количество чисел:", int(result))
```

# Задача 1.10

> Буквы Т, Е, О, Р, И, Я написаны на отдельных карточках.
> Ребёнок берет карточки в случайном порядке и прикладывает одну к другой:
>
> а) 3 карточки;
>
> б) все 6 карточек.
>
> Какова вероятность того, что получится слово:
>
> а) «ТОР»;
>
> б) «ТЕОРИЯ»?

### Решение

### а) Получение слова «ТОР»

Рассмотрим событие $A$ – получение слова «ТОР».

Так как выбираются 3 карточки из 6 и важен порядок их следования, используем размещения без повторений:

$A_n^m=\frac{n!}{(n-m)!}$

Общее число возможных вариантов:

$ n=A_6^3 $

Благоприятный вариант только один:

$ m=1 $

Вероятность события:

$P(A)=\frac{m}{n}=\frac{1}{A_6^3}$

```python
import scipy.special as sc

# количество всех вариантов размещения 3 букв из 6
n = sc.factorial(6) / sc.factorial(6-3)
m = 1
P_A = m / n
print("Вероятность получения слова ТОР:", round(P_A, 4))
```

### б) Получение слова «ТЕОРИЯ»

Рассмотрим событие $B$ – получение слова «ТЕОРИЯ».

Используются все 6 карточек, поэтому возможны перестановки без повторений: 
$P_n=n!$

Общее число вариантов: 
$n=P_6=6!$

Благоприятный вариант только один: 
$m=1$

Вероятность события:
$P(B)=\frac{m}{n}=\frac{1}{P_6}$

```python
import scipy.special as sc

# количество всех перестановок 6 букв
n = sc.factorial(6)
m = 1
P_B = m / n
print("Вероятность получения слова ТЕОРИЯ:", round(P_B, 4))
```

# Задача 1.11

> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС»,
> если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

### Решение

* Рассмотрим событие $B$ - получение слова «АНАНАС».

Всего имеется 6 карточек, поэтому общее число перестановок:

$ n=P_6=6! $

Так как буквы повторяются:
- буква А встречается 3 раза;
- буква Н встречается 2 раза;
- буква С встречается 1 раз.

Используем перестановки с повторениями:

$P_n(n_1,n_2,...,n_k)=\frac{n!}{n_1!\cdot n_2!\cdot ...\cdot n_k!}$

Количество благоприятных случаев:

$m=P_3\cdot P_2$

Вероятность события:

$P(B)=\frac{m}{n}$

```python
import scipy.special as sc

# общее количество перестановок 6 карточек
n = sc.factorial(6)

# благоприятные перестановки:
# три буквы А и две буквы Н
m = sc.factorial(3) * sc.factorial(2)

P_B = m / n

print("Вероятность получения слова АНАНАС:", round(P_B, 4))
```

# Задача 1.16

> За круглым столом рассаживаются 5 мужчин и 5 женщин.
> Найти вероятность того, что:
>
> а) никакие два лица одного пола не сядут рядом;
>
> б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

### Решение

## а) Мужчины и женщины чередуются

Рассмотрим событие $A$ – никакие два человека одного пола не сидят рядом.

Общее число способов рассадки 10 человек за круглым столом:

$ n=P_{10}=10! $

Благоприятные случаи:

Сначала рассадим женщин:

$5!$

Мужчины могут занять оставшиеся места:

$5!$

Также возможно обратное расположение (мужчины на месте женщин):

$ m=2(5!)^2 $

Вероятность события:

$P(A)=\frac{m}{n}$


```python
import scipy.special as sc

n = sc.factorial(10)

m = 2 * sc.factorial(5)**2

P_A = m / n

print("Вероятность события A:", round(P_A, 4))
```

## б) Супруги сидят рядом

Рассмотрим событие $B$ – каждая супружеская пара сидит рядом.

Общее число способов рассадки:

$ n=10! $

Объединим каждую супружескую пару в один блок.

Тогда имеется 5 блоков, которые можно расположить: $5!$


Внутри каждой пары муж и жена могут поменяться местами: $2^5$


Количество благоприятных случаев:

$m=5!2^5$

Вероятность события:

$P(B)=\frac{m}{n}$



```python
import scipy.special as sc

n = sc.factorial(10)

m = sc.factorial(5) * 2**5

P_B = m / n

print("Вероятность события B:", round(P_B, 4))
```

# Задача 1.17

> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов.
> Найти вероятности того, что пассажиры попали:
>
> а) в два купе;
>
> б) в семь купе;
>
> в) в три купе.

### Решение

Общее количество мест в вагоне:

$9\cdot4=36$

Общее число способов выбрать 7 мест из 36:

$n=C_{36}^{7}$


## а) Пассажиры попали в два купе

Рассмотрим событие $A$ – пассажиры попали в два купе.

Выбрать 2 купе из 9 можно:

$C_9^2$

Выбрать 7 мест из 8 мест в этих двух купе:

$C_8^7$

Количество благоприятных случаев:

$m_1=C_9^2\cdot C_8^7$

Вероятность:

$P(A)=\frac{m_1}{n}$


```python
import scipy.special as sc

n = sc.comb(36, 7)

m1 = sc.comb(9, 2) * sc.comb(8, 7)

P_A = m1 / n

print("Вероятность события A:", round(P_A, 4))
```

## б) Пассажиры попали в семь купе

Рассмотрим событие $B$ – пассажиры попали в семь купе.

Выбрать 7 купе из 9:

$C_9^7$

В каждом выбранном купе выбираем по одному месту из четырёх:

$4^7$

Количество благоприятных случаев:

$m_2=C_9^7\cdot4^7$

Вероятность:

$P(B)=\frac{m_2}{n}$


```python
m2 = sc.comb(9, 7) * 4**7

P_B = m2 / n

print("Вероятность события B:", round(P_B, 4))
```

## в) Пассажиры попали в три купе

Рассмотрим событие $D$ – пассажиры попали в три купе.

Выбираем 3 купе из 9:

$C_9^3$

Возможные распределения 7 пассажиров по трём купе:

$4+2+1$

$3+3+1$

$3+2+2$

С учетом перестановок купе:

$6C_4^4C_4^2C_4^1+
3C_4^3C_4^3C_4^1+
3C_4^3C_4^2C_4^2$

Количество благоприятных случаев:

$m_3=C_9^3(6C_4^4C_4^2C_4^1+
3C_4^3C_4^3C_4^1+
3C_4^3C_4^2C_4^2)$

Вероятность:

$P(D)=\frac{m_3}{n}$

```python
m3 = sc.comb(9,3) * (
    6 * sc.comb(4,4) * sc.comb(4,2) * sc.comb(4,1)
    +
    3 * sc.comb(4,3) * sc.comb(4,3) * sc.comb(4,1)
    +
    3 * sc.comb(4,3) * sc.comb(4,2) * sc.comb(4,2)
)

P_D = m3 / n

print("Вероятность события D:", round(P_D, 4))
```
