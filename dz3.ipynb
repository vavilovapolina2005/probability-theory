# Задача

>  В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.
 	
# Классическое решение

```python
from math import comb

total = comb(20, 3) 
answer = 0

for k in range(4):
    p_k = comb(15, k) * comb(5, 3 - k) / total  
    p_white = (6 + k) / 10                 
    answer += p_k * p_white

print("Ответ ", answer)
```

# Моделирование

```python
import random

def one_try():
    urna1 = ["w"] * 15 + ["b"] * 5
    urna2 = ["w"] * 6 + ["b"] * 1

    moved = random.sample(urna1, 3)  
    urna2 += moved                

    ball = random.choice(urna2)    
    return ball == "w"

n = 1000000
count = 0
for i in range(n):
    if one_try():
        count += 1

print("Ответ ", count / n)
```
